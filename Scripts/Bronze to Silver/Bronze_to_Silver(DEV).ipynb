#### Now Extract all the columns from Results string to transform the data

```python
df_silver=df_bronze.select(
        get_json_object(col('results'),"$.gender").alias('Gender'),
        get_json_object(col('results'),"$.name.title").alias('Title'),
        get_json_object(col('results'),"$.name.first").alias('First'),
        get_json_object(col('results'),"$.name.last").alias('Last'),
        concat(get_json_object(col('results'),"$.location.street.number"),lit(', '),get_json_object(col('results'),"$.location.street.name")).alias('Street_Info'),
        get_json_object(col('results'),"$.location.city").alias('City'),
        get_json_object(col('results'),"$.location.state").alias('State'),
        get_json_object(col('results'),"$.location.country").alias('Country'),
        get_json_object(col('results'),"$.location.postcode").alias('Postcode'),
        get_json_object(col('results'),"$.location.coordinates.latitude").alias('Latitude'),
        get_json_object(col('results'),"$.location.coordinates.longitude").alias('Longitude'),
        get_json_object(col('results'),"$.location.timezone.offset").alias('TimeZone_Offset'),
        get_json_object(col('results'),"$.location.timezone.description").alias('TimeZone_Description'),
        get_json_object(col('results'),"$.email").alias('Email'),
        get_json_object(col('results'),"$.login.uuid").alias('userid'),
        get_json_object(col('results'),"$.login.username").alias('User_Name'),
        get_json_object(col('results'),"$.login.password").alias('Password'),
        get_json_object(col('results'),"$.login.salt").alias('Salt'),
        get_json_object(col('results'),"$.login.md5").alias('MD5'),
        get_json_object(col('results'),"$.login.sha1").alias('SHA1'),
        get_json_object(col('results'),"$.login.sha256").alias('SHA256'),
        get_json_object(col('results'),"$.dob.date").alias('Birth_Date'),
        get_json_object(col('results'),"$.dob.age").alias('Age'),
        get_json_object(col('results'),"$.registered.date").alias('Registered_Date'),
        get_json_object(col('results'),"$.registered.age").alias('Registered_Age'),
        get_json_object(col('results'),"$.phone").alias('Phone_Number'),
        get_json_object(col('results'),"$.cell").alias('Cell_Number'),
        get_json_object(col('results'),"$.id.name").alias('ID_Name'),
        get_json_object(col('results'),"$.id.value").alias('ID_Value'),
        get_json_object(col('results'),"$.picture.large").alias('Picture_Large'),
        get_json_object(col('results'),"$.picture.medium").alias('Picture_Medium'),
        get_json_object(col('results'),"$.picture.thumbnail").alias('Picture_thumbnail'),
        get_json_object(col('results'),"$.nat").alias('NAT'),
        col('injestion_timestamp').alias('Injestion_Timestamp')
        )
```

```python
df_valid=df_bronze.filter(trim((col('results')))!='[]')
```

```python
df_normalized=df_valid.withColumn(
    'results_normalized',when(
        trim(col('results')).startswith('['),
        col('results')).otherwise(
            concat(lit("["),col('results'),lit(']')))
        )
```

```python
df_silver1=df_normalized.select(
        get_json_object(col('results_normalized'),"$[0].login.uuid").alias('userid'),
        col('injestion_timestamp').alias('Injestion_Timestamp')
        )
```
